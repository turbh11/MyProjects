```python
import torch
from transformers import pipeline
import json
import glob
import os

device = "cuda:0" if torch.cuda.is_available() else "cpu"
print(f"Starting pipeline on: {device}")

model_id = "ivrit-ai/whisper-large-v3"

transcriber = pipeline(
    "automatic-speech-recognition",
    model=model_id,
    device=device,
    chunk_length_s=30,
    generate_kwargs={"language": "hebrew", "task": "transcribe"}
)

drive_folder = "/content/drive/MyDrive/שמע"
audio_files = glob.glob(os.path.join(drive_folder, "*"))

if not audio_files:
    print("No files found! Check your folder path.")

for audio_path in audio_files:
    # דילוג על קבצי הטקסט וה-JSON שכבר נוצרו
    if audio_path.endswith('.txt') or audio_path.endswith('.json'):
        continue

    base_name = os.path.splitext(audio_path)[0]
    txt_path = f"{base_name}_transcription.txt"
    json_path = f"{base_name}_timestamps.json"

    # הבדיקה שחוסכת זמן: האם הקובץ כבר תומלל?
    if os.path.exists(txt_path):
        print(f"Skipping {audio_path.split('/')[-1]} - Already transcribed!")
        continue

    print(f"\nProcessing {audio_path.split('/')[-1]}...")

    try:
        result = transcriber(audio_path, return_timestamps=True)

        with open(txt_path, "w", encoding="utf-8") as f:
            f.write(result["text"])

        with open(json_path, "w", encoding="utf-8") as f:
            json.dump(result["chunks"], f, ensure_ascii=False, indent=4)

        print(f"Finished! TXT and JSON saved to Drive.")
    except Exception as e:
        print(f"Error processing {audio_path}: {e}")

print("\nAll done!")
```
